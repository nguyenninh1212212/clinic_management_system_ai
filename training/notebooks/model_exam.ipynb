```python
conflict_mapping = {
    "Mỗi khi mang vác đồ nặng, tôi cảm thấy như lưng mình sắp gãy.": "Chấn thương chỉnh hình",
    "Tôi bị phát ban lạ trên cánh tay.": "Da liễu",
    "Tôi bị rụng tóc rất nhiều.": "Da liễu",
    "Tôi cảm thấy tức ngực.": "Hô hấp",
    "Vết thương bị nhiễm trùng của tôi gây ra sốt.": "Ngoại khoa",
    "Đầu tôi đau quá.": "Thần kinh",
}

for text, label in conflict_mapping.items():
    df.loc[df["text"] == text, "label"] = label
```

```python
conflict = df.groupby("text")["label"].nunique()

print("CONFLICT:", (conflict > 1).sum())
```

```python
df_clean = df.drop_duplicates(
    subset=["text", "label"]
).copy()
```
