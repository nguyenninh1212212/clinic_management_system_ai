```python
probabilities = model.predict_proba(test_sentences)

for text, prediction, probs in zip(
    test_sentences,
    predictions,
    probabilities
):
    top_indices = probs.argsort()[-3:][::-1]

    print(f"\n{text}")
    print(f"Prediction: {prediction}")

    for i in top_indices:
        print(
            f"  {model.classes_[i]}: "
            f"{probs[i]:.4f}"
        )
```
