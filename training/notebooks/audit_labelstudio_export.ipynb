```python
import json
import pandas as pd

records = []

for _, row in df.iterrows():
    raw_label = row["label"]

    if pd.isna(raw_label):
        continue

    spans = (
        json.loads(raw_label)
        if isinstance(raw_label, str)
        else raw_label
    )

    for span in spans:
        records.append({
            "id": row["id"],
            "start": span["start"],
            "end": span["end"],
            "text": span["text"],
            "label": ", ".join(span["labels"]),
        })

spans_df = pd.DataFrame(records)

display(spans_df)
```

```python

pd.set_option("display.max_colwidth", None)

label_counts = (
    spans_df["label"]
    .value_counts(dropna=False)
    .rename_axis("label")
    .reset_index(name="count")
)

label_counts["percentage"] = (
    label_counts["count"] / len(spans_df) * 100
).round(2)

display(label_counts)
```

```python
import json
import pandas as pd

audit_records = []

for row_index, row in df.iterrows():
    raw_label = row.get("label")
    source_text = row.get("text")

    if pd.isna(raw_label) or pd.isna(source_text):
        continue

    try:
        spans = (
            json.loads(raw_label)
            if isinstance(raw_label, str)
            else raw_label
        )
    except (json.JSONDecodeError, TypeError):
        print(raw_label)
        audit_records.append({
            "row_index": row_index,
            "id": row.get("id"),
            "issue": "INVALID_JSON",
            "span_text": None,
            "details": "Không parse được cột label",
        })
        continue

    for span in spans:
        start = span.get("start")
        end = span.get("end")
        span_text = span.get("text", "")
        labels = span.get("labels", [])

        if not isinstance(start, int) or not isinstance(end, int):
            issue = "INVALID_OFFSET_TYPE"
            actual_text = None
        elif start < 0 or end > len(source_text) or start >= end:
            issue = "OFFSET_OUT_OF_RANGE"
            actual_text = None
        else:
            actual_text = source_text[start:end]
            issue = (
                None if actual_text == span_text
                else "TEXT_OFFSET_MISMATCH"
            )

        if issue:
            audit_records.append({
                "row_index": row_index,
                "id": row.get("id"),
                "issue": issue,
                "start": start,
                "end": end,
                "span_text": span_text,
                "actual_text": actual_text,
                "labels": labels,
                "details": "Kiểm tra thủ công trước khi sửa",
            })

audit_df = pd.DataFrame(audit_records)

print(f"Tổng số lỗi phát hiện: {len(audit_df)}")
display(audit_df)
```

```python
overlap_records = []

for row_index, row in df.iterrows():
    raw_label = row.get("label")
    source_text = row.get("text")

    if pd.isna(raw_label) or pd.isna(source_text):
        continue

    try:
        spans = (
            json.loads(raw_label)
            if isinstance(raw_label, str)
            else raw_label
        )
    except (json.JSONDecodeError, TypeError):
        continue

    spans = sorted(
        spans,
        key=lambda s: (s.get("start", -1), s.get("end", -1))
    )

    for i, first in enumerate(spans):
        for second in spans[i + 1:]:
            if second.get("start", -1) >= first.get("end", -1):
                break

            overlap_records.append({
                "row_index": row_index,
                "id": row.get("id"),
                "text": source_text,
                "span_1": first.get("text"),
                "label_1": first.get("labels"),
                "start_1": first.get("start"),
                "end_1": first.get("end"),
                "span_2": second.get("text"),
                "label_2": second.get("labels"),
                "start_2": second.get("start"),
                "end_2": second.get("end"),
            })

overlap_df = pd.DataFrame(overlap_records)

print(f"Tổng số cặp span chồng lấn: {len(overlap_df)}")
display(overlap_df)
```
