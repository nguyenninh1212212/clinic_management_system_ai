```python
import pandas as pd

path = (
    "../dataset/disease/"
    "disease_knowledge_base.csv"
)

df = pd.read_csv(path)

diseases = sorted(
    df["disease"]
    .dropna()
    .unique()
)

print("DISEASES")
print("=" * 70)

for index, disease in enumerate(diseases, start=1):
    print(f"{index:3}. {disease}")

print("=" * 70)
print(f"Total diseases: {len(diseases)}")
```

```python
disease_df = pd.DataFrame({
    "id": range(1, len(diseases) + 1),
    "disease": diseases,
})

disease_df
```

```python
disease_symptom_count = (
    df.groupby("disease")["symptom"]
    .nunique()
    .sort_values(ascending=False)
    .reset_index(name="symptom_count")
)

disease_symptom_count
```

```python
def suggest_department(
    disease,
    department_keywords,
    department_priority,
    explicit_department_mapping,
):
    disease_lower = disease.lower()

    # 1. Explicit mapping
    if disease_lower in explicit_department_mapping:
        return [{
            "department": explicit_department_mapping[disease_lower],
            "keywords": ["explicit_mapping"],
        }]

    # 2. Priority rules
    priority_matches = []

    for department, keywords in department_priority.items():
        matched_keywords = [
            keyword
            for keyword in keywords
            if keyword in disease_lower
        ]

        if matched_keywords:
            priority_matches.append({
                "department": department,
                "keywords": matched_keywords,
            })

    if len(priority_matches) == 1:
        return priority_matches

    # 3. Normal keyword rules
    matches = []

    for department, keywords in department_keywords.items():
        matched_keywords = [
            keyword
            for keyword in keywords
            if keyword in disease_lower
        ]

        if matched_keywords:
            matches.append({
                "department": department,
                "keywords": matched_keywords,
            })

    return matches

mapping_rows = []

for disease in disease_list:
    matches = suggest_department(
        disease,
        department_keywords,
        department_priority,
        explicit_department_mapping,
)
    if len(matches) == 1:
        department = matches[0]["department"]
        status = "auto"

    elif len(matches) > 1:
        department = None
        status = "review"

    else:
        department = None
        status = "unmapped"

    mapping_rows.append({
        "disease": disease,
        "department": department,
        "status": status,
        "matches": matches,
    })

mapping_df = pd.DataFrame(mapping_rows)

mapping_df.head(20)
```
