```python
all_labels = set()

for split in [
    train_dataset,
    validation_dataset,
    test_dataset
]:
    for labels in split["labels"]:
        all_labels.update(labels)

label_names = sorted(all_labels)

print("Number of labels:", len(label_names))
print(label_names)
```

```python
label2id = {
    label: i
    for i, label in enumerate(label_names)
}

id2label = {
    i: label
    for i, label in enumerate(label_names)
}

print(label2id)
print(id2label)
```

```python
from transformers import AutoTokenizer

MODEL_NAME = "vinai/phobert-base"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=False,
)

print("Tokenizer:", type(tokenizer))
print("Fast:", tokenizer.is_fast)
print("CLS:", tokenizer.cls_token)
print("SEP:", tokenizer.sep_token)
print("PAD:", tokenizer.pad_token)
```

```python
MAX_LENGTH = 256


def tokenize_function(example):
    input_ids = []
    attention_mask = []
    aligned_labels = []

    words = example["words"]
    labels = example["labels"]

    for word, label in zip(words, labels):

        tokens = tokenizer(
            word,
            add_special_tokens=False,
        )

        word_input_ids = tokens["input_ids"]

        if not word_input_ids:
            continue

        # Token đầu tiên nhận label
        input_ids.append(word_input_ids[0])
        attention_mask.append(1)
        aligned_labels.append(label2id[label])

        # Các sub-token còn lại bỏ qua loss
        for token_id in word_input_ids[1:]:
            input_ids.append(token_id)
            attention_mask.append(1)
            aligned_labels.append(-100)

    # CLS
    input_ids = (
        [tokenizer.cls_token_id]
        + input_ids
        + [tokenizer.sep_token_id]
    )

    attention_mask = (
        [1]
        + attention_mask
        + [1]
    )

    aligned_labels = (
        [-100]
        + aligned_labels
        + [-100]
    )

    # Truncate
    input_ids = input_ids[:MAX_LENGTH]
    attention_mask = attention_mask[:MAX_LENGTH]
    aligned_labels = aligned_labels[:MAX_LENGTH]

    return {
        "input_ids": input_ids,
        "attention_mask": attention_mask,
        "labels": aligned_labels,
    }
```

```python
train_tokenized = train_dataset.map(
    tokenize_function,
    remove_columns=train_dataset.column_names,
)

validation_tokenized = validation_dataset.map(
    tokenize_function,
    remove_columns=validation_dataset.column_names,
)

test_tokenized = test_dataset.map(
    tokenize_function,
    remove_columns=test_dataset.column_names,
)
```

# Continued fine-tuning: NEGATION (37 → 39 nhãn)

Chạy cell mount Drive (cell 6), sau đó chạy **chỉ các cell của phần bổ sung này**. Không chạy lại phần train PhoBERT gốc.

Upload `negation_3000_annotated_DRAFT.jsonl` vào Google Drive và đổi tên thành `clinic_ai/negation_annotated.jsonl`. Đây là dữ liệu **gán nhãn tự động chưa kiểm duyệt**. Code sẽ loại các dòng có span không khớp ranh giới từ; không tự sửa nhãn.

Cần có checkpoint cũ `clinic_ai/symptom_ner_model/final` và dataset `clinic_ai/vietmed_ner`.


```python
import os, json, re, random, torch
from datasets import load_from_disk, Dataset, concatenate_datasets
from transformers import (AutoTokenizer, AutoModelForTokenClassification,
    DataCollatorForTokenClassification, TrainingArguments, Trainer)

BASE_DIR = "/content/drive/MyDrive/clinic_ai"
OLD_MODEL_DIR = f"{BASE_DIR}/symptom_ner_model/final"
OLD_DATA_DIR = f"{BASE_DIR}/vietmed_ner"
NEGATION_JSONL = f"{BASE_DIR}/negation_annotated.jsonl"
NEW_OUTPUT_DIR = f"{BASE_DIR}/symptom_ner_negation"
MAX_LENGTH = 256

assert os.path.isdir(OLD_MODEL_DIR), f"Không tìm thấy checkpoint: {OLD_MODEL_DIR}"
assert os.path.isfile(NEGATION_JSONL), (
    f"Chưa có dữ liệu đã gán nhãn: {NEGATION_JSONL}. "
    "Hãy export annotation từ Label Studio rồi chuyển sang JSONL trước khi chạy."
)

old_model = AutoModelForTokenClassification.from_pretrained(OLD_MODEL_DIR)
tokenizer_neg = AutoTokenizer.from_pretrained(OLD_MODEL_DIR, use_fast=False)
old_id2label = {int(k): v for k, v in old_model.config.id2label.items()}
old_label2id = {v: k for k, v in old_id2label.items()}
assert sorted(old_id2label) == list(range(len(old_id2label))), "Old label IDs không liên tục"
new_label2id = dict(old_label2id)
for label in ("B-NEGATION", "I-NEGATION"):
    if label not in new_label2id:
        new_label2id[label] = len(new_label2id)
new_id2label = {i: label for label, i in new_label2id.items()}
print("Nhãn cũ:", len(old_label2id), "Nhãn mới:", len(new_label2id))
print("Nhãn phủ định:", {k:v for k,v in new_label2id.items() if 'NEGATION' in k})
```

```python
# Tạo model với head lớn hơn; sao chép đúng các hàng classifier cũ.
model_neg = AutoModelForTokenClassification.from_pretrained(
    OLD_MODEL_DIR, num_labels=len(new_label2id),
    label2id=new_label2id, id2label=new_id2label,
    ignore_mismatched_sizes=True,
)
assert hasattr(old_model, 'classifier') and hasattr(model_neg, 'classifier')
with torch.no_grad():
    for label, old_idx in old_label2id.items():
        new_idx = new_label2id[label]
        model_neg.classifier.weight[new_idx].copy_(old_model.classifier.weight[old_idx])
        model_neg.classifier.bias[new_idx].copy_(old_model.classifier.bias[old_idx])
print('Đã giữ encoder và sao chép các lớp classifier cũ; nhãn mới chưa được học.')
```

```python
# Đọc dữ liệu, kiểm tra và loại các dòng annotation không khớp biên từ.
# Dataset là dữ liệu tổng hợp DRAFT; cần kiểm duyệt trước khi dùng production.
with open(NEGATION_JSONL, encoding='utf-8-sig') as f:
    rows = [json.loads(line) for line in f if line.strip()]

def validate_row(row):
    text = row['text']
    spans = sorted(row['entities'], key=lambda x: x['start'])
    words = list(re.finditer(r'\S+', text))
    previous_end = 0
    for ent in spans:
        a, b, label = ent['start'], ent['end'], ent['label']
        if not (0 <= a < b <= len(text) and a >= previous_end):
            return False
        if f'B-{label}' not in new_label2id or f'I-{label}' not in new_label2id:
            return False
        touched = [w for w in words if w.start() < b and w.end() > a]
        if not touched or touched[0].start() != a or touched[-1].end() != b:
            return False
        previous_end = b
    return True

valid_rows, rejected_rows = [], []
seen = set()
for row in rows:
    if row['text'] in seen or not validate_row(row):
        rejected_rows.append(row)
    else:
        valid_rows.append(row)
        seen.add(row['text'])

print('Tổng:', len(rows), 'Hợp lệ:', len(valid_rows), 'Loại:', len(rejected_rows))
assert len(valid_rows) >= 20, 'Không đủ dữ liệu hợp lệ'
assert any(e['label'] == 'NEGATION' for r in valid_rows for e in r['entities'])

# Chia trước khi tokenize, giữ seed để tái lập.
random.Random(42).shuffle(valid_rows)
n_val = max(1, round(len(valid_rows) * 0.2))
new_train_rows = valid_rows[n_val:]
new_val_rows = valid_rows[:n_val]
print('New train:', len(new_train_rows), 'New validation:', len(new_val_rows))
```

```python
# PhoBERT tokenizer slow: căn chỉnh bằng whitespace token và kiểm tra span biên.
# Các span cần khớp với biên của token theo cách tách từ này.
# Với tiếng Việt, nếu muốn nhận diện đa âm tiết chính xác hơn, nên chuẩn hóa
# cùng một chiến lược word segmentation cho train và inference.
def tokenize_annotated(row):
    text = row['text']
    matches = list(re.finditer(r'\S+', text))
    spans = sorted(row['entities'], key=lambda x: x['start'])
    word_labels = ['0' if '0' in new_label2id else 'O'] * len(matches)
    for ent in spans:
        touched = [i for i,m in enumerate(matches)
                   if m.start() < ent['end'] and m.end() > ent['start']]
        assert touched, (text, ent)
        # Không được âm thầm mở rộng annotation sang token bên cạnh.
        assert matches[touched[0]].start() == ent['start'] and matches[touched[-1]].end() == ent['end'], (text,ent)
        for j,i in enumerate(touched):
            word_labels[i] = ('B-' if j == 0 else 'I-') + ent['label']
    ids, labels = [], []
    for match, lab in zip(matches, word_labels):
        subids = tokenizer_neg(match.group(), add_special_tokens=False)['input_ids']
        if not subids: continue
        ids.extend(subids)
        labels.extend([new_label2id[lab]] + [-100]*(len(subids)-1))
    ids = [tokenizer_neg.cls_token_id] + ids[:MAX_LENGTH-2] + [tokenizer_neg.sep_token_id]
    labels = [-100] + labels[:MAX_LENGTH-2] + [-100]
    return {'input_ids':ids, 'attention_mask':[1]*len(ids), 'labels':labels}

new_train = Dataset.from_list([tokenize_annotated(r) for r in new_train_rows])
new_val = Dataset.from_list([tokenize_annotated(r) for r in new_val_rows])
print(new_train[0])
```

```python
# Trộn dữ liệu VietMed-NER cũ để hạn chế catastrophic forgetting.
old_ds = load_from_disk(OLD_DATA_DIR)

def tokenize_old(row):
    ids, labels = [], []
    for word, label in zip(row['words'], row['labels']):
        subids = tokenizer_neg(str(word), add_special_tokens=False)['input_ids']
        if not subids: continue
        assert label in new_label2id, f'Nhãn cũ không có trong model: {label}'
        ids.extend(subids)
        labels.extend([new_label2id[label]] + [-100]*(len(subids)-1))
    ids = [tokenizer_neg.cls_token_id] + ids[:MAX_LENGTH-2] + [tokenizer_neg.sep_token_id]
    labels = [-100] + labels[:MAX_LENGTH-2] + [-100]
    return {'input_ids':ids, 'attention_mask':[1]*len(ids), 'labels':labels}

old_train = old_ds['train'].map(tokenize_old, remove_columns=old_ds['train'].column_names)
old_val = old_ds['validation'].map(tokenize_old, remove_columns=old_ds['validation'].column_names)
# Giới hạn old data để tránh NEGATION bị chìm trong quá nhiều câu không có nhãn mới.
old_train = old_train.shuffle(seed=42).select(range(min(len(old_train), max(100, len(new_train)*2))))
train_combined = concatenate_datasets([old_train, new_train]).shuffle(seed=42)
# Validation riêng: old và new, không gộp metric để tránh che lấp NEGATION.
print('Combined train:',len(train_combined), 'Old val:',len(old_val), 'New val:',len(new_val))
```

```python
# Huấn luyện tiếp từ checkpoint cũ; lưu sang thư mục mới.
args_neg = TrainingArguments(
    output_dir=NEW_OUTPUT_DIR, eval_strategy='epoch', save_strategy='epoch',
    learning_rate=1e-5, num_train_epochs=3,
    per_device_train_batch_size=8, per_device_eval_batch_size=8,
    weight_decay=0.01, logging_steps=20, save_total_limit=2,
    load_best_model_at_end=True, metric_for_best_model='eval_loss',
    greater_is_better=False, report_to='none',
    fp16=torch.cuda.is_available(),
)
trainer_neg = Trainer(
    model=model_neg, args=args_neg, train_dataset=train_combined,
    eval_dataset=new_val,
    data_collator=DataCollatorForTokenClassification(tokenizer_neg),
    processing_class=tokenizer_neg,
)
trainer_neg.train()
print('New NEGATION validation:', trainer_neg.evaluate(new_val))
print('Old VietMed validation:', trainer_neg.evaluate(old_val))
NEW_FINAL_DIR = f'{NEW_OUTPUT_DIR}/final'
trainer_neg.save_model(NEW_FINAL_DIR)
tokenizer_neg.save_pretrained(NEW_FINAL_DIR)
print('Saved:', NEW_FINAL_DIR)
```

```python
# Đánh giá Precision/Recall/F1 theo nhãn trên validation tổng hợp.
# Chỉ là chỉ số kiểm tra kỹ thuật; cần thêm tập test thực tế độc lập.
import numpy as np
from seqeval.metrics import classification_report, precision_score, recall_score, f1_score

pred = trainer_neg.predict(new_val)
pred_ids = np.argmax(pred.predictions, axis=-1)
true_sequences, predicted_sequences = [], []
for ids, labels in zip(pred_ids, pred.label_ids):
    true_sequences.append([new_id2label[int(y)] for y in labels if y != -100])
    predicted_sequences.append([new_id2label[int(p)] for p, y in zip(ids, labels) if y != -100])
print(classification_report(true_sequences, predicted_sequences, digits=4, zero_division=0))
print('Micro F1:', f1_score(true_sequences, predicted_sequences))
```

## Sau training

- Đo Precision/Recall/F1 **theo entity** trên tập test độc lập cho `NEGATION` và các nhãn NER cũ (cell trên mới báo loss, **chưa đủ** để kết luận model tốt).
- Tập JSONL mới phải được annotate đầy đủ; không tự động coi triệu chứng chưa annotate là `O`.
- Ví dụ `Tôi không đau đầu nhưng chóng mặt`: NER chỉ trả về `không`, `đau đầu`, `chóng mặt`; cần bước xác định scope để suy ra đau đầu = absent, chóng mặt = present.
- **Trạng thái:** đã chuẩn bị notebook mở rộng nhãn và training; chưa có annotation JSONL nên chưa thể train hoặc xác nhận F1.
